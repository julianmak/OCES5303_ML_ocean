### General spiel about assessment

***Your hand in should be in the form of a Jupyter notebook, and no other form of hand-ins will be accepted***. The use of Jupyter notebook and its Python component is part of the assessment criteria for the *presentation* and *coding* portion. Hand these in through Canvas in the usual way. You are graded on the following attributes:

1) **scientific content** (40%)

2) **writing, presentation and referencing** (30%)

3) **use of Jupyter and/or Python coding** (30%)

4) **orginality** (10%; analysis beyond scope of course, use of memes and puns, reproducible code that runs fresh; surprise me)

See the samples assignments I've made for the kind of things we might be expecting. We will probably be fairly loose with giving credit. 50% or below would count as unsatisfactory, while around 85% or above would be an A grade.

You are allowed to use other Python packages if you find them, but see point b) below.

a) ***Late assignments get a penalty of 1% of full marks per minute*** (so don't bother handing in anything after 2 hours). We will still mark it and give feedback, but you just don't get the credit. Excuses could be entertained but you will need sufficient evidence to back this up (e.g. your internet went down in the area and you have some pictorial/written demosntration for this).

b) ***You are to hand in a rendered Jupyter notebook and that is the only thing that will be marked***, and no other formats or additional files will be taken into account of; we will turn your notebooks into another format to do the marking on. The versions that get marked will be the ones we render, not whatever you see on your laptop; if your notebook for example uses local files as part of the text for example that does not exist for us (because we are not going to accept anything other than the notebook), then it is on you to adjust with your workflow.

c) ***There is a penalty for excessive length***. You can try rendering your own HTML to see what it looks like, and if the result is more than 15-20 pages that is too long, which will hit category 2 above (because it's too long), and maybe category 3 above (because your code is too long and/or you are not suppressing unncessary outputs).

d) ***Originality (category 4 above) is extra and treat it as such***. Cutting-edge things done poorly would still be evaluated as "poor"; you may get some marks for category 4, but you could lose marks in all other parts. 

If you upload your assignment onto a GitHub repository and give us the link, with the appropriate things built in so that we can open it in Colab, then you get extra credit because of making the things openly reproducible and helping us somewhat. Again, it needs to be done "well", not just "done"; the latter gives you no credit by itself

e) ***Plagiarism***: By all means consult each other and/or work together, but the files you hand in should be done and written up separately. **The default for anyone accused with plagiarism is ZERO on the assignment**, and depending on whether you decide to contest and the result of the appeal, possibly lead to an official note of plagiarism on your transcript (arguing has a risk associated with it but it is part of the system).

A few things count as plagiarism:

**Copying between students, and the default is that ALL parties involved get zero for the assignment**, regardless of whether the side can demonstrate they were copied from (extra incentive to keep the writing separate).

**Copying text without citation is plagiarism**. Use quotation marks and give reference if you are directly lifting text, but don't do this too often (and will result in text looking cluttered, and not getting full credit for the *presentation* aspect).

**Code is a slightly more grey area**, but I will just say no one has ever really been punished for being cautious and generous with citations, but make sure you present it well (e.g. overburdening text with citations will make the presentation ugly, and will not get full credit for the *presentation* aspect say).

If we are making a penalty on plagiarism it probably means we think we have a sufficiently strong case that is unlikely arguable (I don't fight battles I know I will lose).

# Assessment 1 (15% of total course grade)

For this assignment you have a choice of either using the Argo data (`argo_data_scrap.ipynb`), bacteria data (`bacteria_data_scrap.ipynb`) or some numerical model data (`model_data_scrap.ipynb`), up to you, but do not use other choices of data (you have that option in the extended project later). Pros and cons are that:

* Argo data has substantially fewer variables (physics focused), but more effort might be needed to load the data.
* Bacteria data is easy to load, but has a ton of variables (biogeochemistry focused).
* Numerical model data behaviour depends on the parameters you choose.

With any of those choices you should demonstrate the ability to deploy some things we have gone through in sessions 1 to 3, i.e. those things that live in `sklearn`. Up to you what you choose to deploy, but if you do something you should justify why you are doing it with a scientific/practical reason (and comment accordingly on what happens if you don't do it).

You can get high marks by doing few things but all really well, or doing lots of things reasonably enough, there is no one way that is "right" (just like there is no single way of something being a "good" article, although there are usually commonalities between "good" things). I would recommend doing two to three models, doing one single task (classification or regression), and being reasonably thorough with model comparison, sensitivity on hyper-parameter exploration, model validation and so forth. Your document should include the following:

a) Describe what the data includes, and explain why one might want to do regression and/or classification with it [*be able to do some background research*]

b) Perform regression and/or classfication on the data with some methods (along with all the other bits that come with it, e.g. describing the method) [*practise and demonstrate using Python to do some scientific computation*]

c) Write some of these things up and describe them using the Markdown cells [*practise and demonstrate understanding of Jupyter notebooks*]

d) Any others that could fall under originality (memes welcome, but scientific content should always come first rather than just random content for the hell of it)

See the relevant `*_scrap.ipynb` notebooks for suggested things to do with the data (I've kept the descriptions quite generic).
